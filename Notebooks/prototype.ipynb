```python
import cv2
import matplotlib.pyplot as plt
```

```python
image_path="../datasets/sample.jpg"
image=cv2.imread(image_path)
if image is not None:
    print(f"Success! Image loaded. Dimensions : {image.shape}")
    plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
    plt.axis('off')
    plt.show()
else:
    print("Image file not found yet. Make sure to drop a sample into your folder!")
```
